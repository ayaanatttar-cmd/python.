```python
print("--- 1. LIST OPERATIONS ---")
my_list = ["apple", "banana", "cherry"]
print("Initial List:", my_list)
my_list.append("orange")
print("1. After append('orange'):", my_list)
my_list.insert(1, "mango")
print("2. After insert(1, 'mango'):", my_list)
my_list[0] = "grapes"
print("3. After updating index 0 to 'grapes':", my_list)
my_list.remove("banana")
print("4. After remove('banana'):", my_list)
popped_item = my_list.pop(2)
print(f"5. After pop(2) [removed '{popped_item}']:", my_list)



print("--- 2. TUPLE OPERATIONS ---")
my_tuple = (10, 20, 30, 20, 40)
print("Initial Tuple:", my_tuple)
print("1. Element at index 2:", my_tuple[2])
print("2. Count of number 20 in tuple:", my_tuple.count(20))
print("3. Index of value 30:", my_tuple.index(30))
new_tuple = my_tuple + (50, 60)
print("4. After concatenating (50, 60):", new_tuple)
temp_list = list(my_tuple)
temp_list.remove(20)  
my_tuple = tuple(temp_list)
print("5. After converting to list, removing 20, and converting back:", my_tuple)



print("--- 3. DICTIONARY OPERATIONS ---")
my_dict = {"name": "Alice", "age": 20, "city": "New York"}
print("Initial Dictionary:", my_dict)
print("1. Value for key 'name':", my_dict["name"])
my_dict["course"] = "Python"
print("2. After adding 'course': 'Python':", my_dict)
my_dict["age"] = 21
print("3. After updating 'age' to 21:", my_dict)
removed_val = my_dict.pop("city")
print(f"4. After pop('city') [removed '{removed_val}']:", my_dict)
del my_dict["course"]
print("5. After 'del my_dict[\"course\"]':", my_dict)
```
