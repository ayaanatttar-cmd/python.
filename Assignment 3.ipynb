```python
def triangle(a, b, c):
    
    sides = sorted([a, b, c])
    a, b, c = sides  
    return a**2 + b**2 == c**2



x = float(input("Enter side 1: "))
y = float(input("Enter side 2: "))
z = float(input("Enter side 3: "))

if triangle(x, y, z):
    print("Right-angled triangle")
else:
    print("Not a right-angled triangle")
    
```
