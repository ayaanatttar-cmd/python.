```python
import pandas as pd
import numpy as np


sample_data = {
    'StudentID': [101, 102, 103, 103],
    'Name': ['Aarav', 'Bhavna', 'Chetan', 'Chetan'],
    'Marks': [85.0, np.nan, 92.0, 92.0],
    'City': ['Mumbai', 'Delhi', np.nan, 'Nashik']
}


df_sample = pd.DataFrame(sample_data)
df_sample.to_csv('students_4_entries.csv', index=False)

print("--- 1. Loading Initial Dataset (4 Entries) ---")
df = pd.read_csv('students_4_entries.csv')
print(df)
print("\n" + "="*50 + "\n")


print("--- 2. Dataset Information ---")
print("\nDataset Info:")
df.info()

print("\nStatistical Summary:")
print(df.describe())
print("\n" + "="*50 + "\n")


print("--- 3. Handling Missing Values ---")
print("\nMissing values count per column:")
print(df.isnull().sum())


marks_mean = df['Marks'].mean()
df['Marks'] = df['Marks'].fillna(marks_mean)


city_mode = df['City'].mode()[0]
df['City'] = df['City'].fillna(city_mode)

print("\nDataFrame after filling missing values:")
print(df)
print("\n" + "="*50 + "\n")


print("--- 4. Handling Duplicate Records ---")

df = df.drop_duplicates(subset=['StudentID', 'Name'], keep='first')

print("\nDataFrame after removing duplicate entries:")
print(df)
print("\n" + "="*50 + "\n")


print("--- 5. Filtering Data using loc[] ---")

high_scorers = df.loc[df['Marks'] >= 85, ['StudentID', 'Name', 'Marks', 'City']]

print("\nStudents scoring 85 marks or higher:")
print(high_scorers)
print("\n" + "="*50 + "\n")


print("--- 6. Final Cleaned Output ---")
print(df)
```
